```python
import pandas as pd

df = pd.read_csv("data_kantin.csv")
```

```python
df.head()
df.info()
df.describe()
df.shape
```

```python
df.isna().sum()          # cek missing value per kolom
df.duplicated().sum()    # cek jumlah baris duplikat
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df.groupby('harga')['menu'].unique()
```

```python
harga_ke_menu = df.dropna(subset=['menu']).groupby('harga')['menu'].agg(lambda x: x.mode()[0])
mask_menu_kosong = df['menu'].isna()
df.loc[mask_menu_kosong, 'menu'] = df.loc[mask_menu_kosong, 'harga'].map(harga_ke_menu)
```

```python
df['terjual'] = df['terjual'].fillna(df.groupby('menu')['terjual'].transform('median'))
```

```python
df['tanggal'] = pd.to_datetime(df['tanggal'])   # dari teks jadi tanggal asli
df['terjual'] = df['terjual'].round().astype(int)  # dari desimal (hasil median) jadi bilangan bulat
```

```python
df.isna().sum()       # harus 0 semua
df.duplicated().sum() # harus 0
df.dtypes             # cek semua tipe sudah benar
```

```python
print(df.isna().sum())
print(df.duplicated().sum())
print(df.dtypes)
```

```python
df['pendapatan'] = df['harga'] * df['terjual']
```

```python
penjualan_tinggi = df[df['terjual'] >= 25]
```

```python
df_sorted = df.sort_values('pendapatan', ascending=False)
```

```python
ringkasan_kategori = df.groupby('kategori').agg(
    total_terjual=('terjual', 'sum'),
    total_pendapatan=('pendapatan', 'sum')
).sort_values('total_pendapatan', ascending=False)
```
